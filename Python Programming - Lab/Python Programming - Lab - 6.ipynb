### 01) WAP to find sum of tuple elements.

```python
t=(1,2,3,4)
print('sum of the tuple :',sum(t))
```

### 02) WAP to find Maximum and Minimum K elements in a given tuple.

```python
t =(1,2,4,6,34,6,43,56,324,5,566)
k = int(input('Enter value of K :'))
temp=[]
for i in t:
    if temp.count(i)==0:
        temp.append(i)
temp = tuple(temp)
print('Minimum',k,'elements:',tuple(sorted(temp)[:k]))
print('Maximum',k,'elements:',tuple(sorted(temp,reverse=True)[:k]))
```

### 03) WAP to find tuples which have all elements divisible by K from a list of tuples.

```python
li =[(1,2,4),(6,34,6),(45,56,324),(5,566)]
temp=[]
k=int(input('Enter value of k:'))
for i in li:
    isValid = True
    for j in i:
        if j%k!=0:
            isValid=False
    if isValid:
        temp.append(i)
print(temp)
```

### 04) WAP to create a list of tuples from given list having number and its cube in each tuple.

```python
li=[1,3,65,23,7,3,6,2,7,3,5,8]
t=[(i,i**3)for i in li]
print(t)
```

### 05) WAP to find tuples with all positive elements from the given list of tuples.

```python
li =[(1,2,4),(6,-34,6),(45,56,324),(5,-566)]
temp=[]
for i in li:
    isValid = True
    for j in i:
         if j<0:
            isValid = False
    if isValid:
        temp.append(i)
print(temp)
```

### 07) WAP to remove tuples of length K.

```python
li =[(1,2,4),(6,-34,6),(45,56,324),(5,-566)]
k = int(input('Enter length of tuple to remove:'))
temp =[]
for i in li:
    if len(i)!=k:
        temp.append(i)
print(temp)
```

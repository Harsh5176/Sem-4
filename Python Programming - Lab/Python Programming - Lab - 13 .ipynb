### 02) Create a class named Bank_Account with Account_No, User_Name, Email,Account_Type and Account_Balance data members. Also create a method GetAccountDetails() and DisplayAccountDetails(). Create main method to demonstrate the Bank_Account class.

```python
class Bank_Account:
    def __init__(self, account_no, user_name, email, account_type, account_balance):
        """Initialize bank account details"""
        self.account_no = account_no
        self.user_name = user_name
        self.email = email
        self.account_type = account_type
        self.account_balance = account_balance

    def GetAccountDetails(self):
        """Accept account details from the user"""
        self.account_no = input("Enter Account Number: ")
        self.user_name = input("Enter User Name: ")
        self.email = input("Enter Email: ")
        self.account_type = input("Enter Account Type (Savings/Current): ")
        self.account_balance = float(input("Enter Account Balance: "))

    def DisplayAccountDetails(self):
        """Display the account details"""
        print("\n--- Account Details ---")
        print(f"Account Number: {self.account_no}")
        print(f"User Name: {self.user_name}")
        print(f"Email: {self.email}")
        print(f"Account Type: {self.account_type}")
        print(f"Account Balance: {self.account_balance:.2f}")

# Main function to demonstrate the class
if __name__ == "__main__":
    # Creating an object of Bank_Account with some initial values
    account1 = Bank_Account("123456789", "Alice", "alice@example.com", "Savings", 5000.0)

    # Displaying initial account details
    account1.DisplayAccountDetails()

    # Accepting new account details from the user
    print("\nEnter New Account Details:")
    account1.GetAccountDetails()

    # Displaying updated account details
    account1.DisplayAccountDetails()

```

### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
import math

class Circle:
    def __init__(self, radius):
        """Initialize Cricle attributes"""
        self.radius = radius 

    def area(self):
        return math.pi * self.radius**2

    def perimeter(self):
        return 2 * math.pi * self.radius 

if __name__ == "__main__":
    r = float(input("Enter the radius of the circle: "))
    Circle = Circle(r)

print(f"\nArea of Circle: {Circle.area():.2f}")    
print(f"Perimeter of Circle: {Circle.perimeter():.2f}")
```

### 04) Create a class for employees that includes attributes such as name, age, salary, and methods to update and display employee information.

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class BankAccount:
    def __init__(self, account_holder, balance=0.0):
        self.account_holder = account_holder
        self.balance = balance
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New balance: ${self.balance}")
        else:
            print("Deposit amount must be positive.")
    
    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.balance}")
        elif amount > self.balance:
            print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")
    
    def check_balance(self):
        print(f"Account balance for {self.account_holder}: ${self.balance}")

# Example usage:
account = BankAccount("John Doe", 1000)
account.deposit(500)
account.withdraw(200)
account.check_balance()



```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class InventoryItem:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def add_stock(self, amount):
        if amount > 0:
            self.quantity += amount
            print(f"Added {amount} of {self.name}. New quantity: {self.quantity}")
        else:
            print("Invalid amount to add.")

    def remove_stock(self, amount):
        if 0 < amount <= self.quantity:
            self.quantity -= amount
            print(f"Removed {amount} of {self.name}. New quantity: {self.quantity}")
        elif amount > self.quantity:
            print("Not enough stock available.")
        else:
            print("Invalid amount to remove.")

    def update_price(self, new_price):
        if new_price > 0:
            self.price = new_price
            print(f"Updated price of {self.name} to ${self.price}")
        else:
            print("Invalid price.")

    def display_info(self):
        print(f"Item: {self.name}, Price: ${self.price}, Quantity: {self.quantity}")

# Example usage
item = InventoryItem("Laptop", 1000, 5)
item.add_stock(3)
item.remove_stock(2)
item.update_price(1200)
item.display_info()
```

### 07) Create a Class with instance attributes of your choice.

```python
class Car:
    def __init__(self, brand, model, year, mileage=0):
        self.brand = brand
        self.model = model
        self.year = year
        self.mileage = mileage

    def drive(self, miles):
        if miles > 0:
            self.mileage += miles
            print(f"Driven {miles} miles. Total mileage: {self.mileage} miles.")
        else:
            print("Miles must be positive.")

    def display_info(self):
        print(f"Car: {self.brand} {self.model}, Year: {self.year}, Mileage: {self.mileage} miles")

# Example usage
my_car = Car("Toyota", "Camry", 2022)
my_car.drive(100)
my_car.display_info()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute

    def add_time(self, other_time):
        total_minutes = self.minute + other_time.minute
        extra_hours = total_minutes // 60
        new_minutes = total_minutes % 60

        total_hours = self.hour + other_time.hour + extra_hours
        new_hours = total_hours % 24  # Keeps it within 24-hour format

        return Time(new_hours, new_minutes)

    def display(self):
        print(f"Time: {self.hour:02d}:{self.minute:02d}")

# Example usage
time1 = Time(5, 45)
time2 = Time(2, 30)

result_time = time1.add_time(time2)
result_time.display()
```

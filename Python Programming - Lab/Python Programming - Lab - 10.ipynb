# Exception Handling

### 01) WAP to handle following exceptions:
1. ZeroDivisionError
2. ValueError
3. TypeError
#### Note: handle them using separate except blocks and also using single except block too.

```python
# Handling separately
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    result = a / b  # May cause ZeroDivisionError
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input! Please enter a valid integer.")
except TypeError:
    print("Error: Unsupported operation due to type mismatch.")

# Handling in a single block
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    result = a / b
    print("Result:", result)
except (ZeroDivisionError, ValueError, TypeError) as e:
    print(f"Error occurred: {e}")


```

### 02)  WAP to handle following exceptions:
1. IndexError
2. KeyError

```python
# Handling IndexError
try:
    lst = [1, 2, 3]
    print(lst[5])  # Accessing an invalid index
except IndexError:
    print("Error: List index out of range.")

# Handling KeyError
try:
    d = {"name": "Alice"}
    print(d["age"])  # Accessing a non-existing key
except KeyError:
    print("Error: Key not found in dictionary.")
```

### 04) WAP that catches all type of exceptions in a single except block.

```python
# Handling RecursionError
def recursive_function():
    return recursive_function()  # Infinite recursion

try:
    recursive_function()
except RecursionError:
    print("Error: Recursion limit exceeded!")

# Handling MemoryError
try:
    large_list = [1] * (10**10)  # Attempting to create a huge list
except MemoryError:
    print("Error: Memory limit exceeded!")
```

### 05) WAP to demonstrate else  and finally block.

```python
try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))
    result = num1 / num2
except ZeroDivisionError:
    print("Error: Division by zero!")
except ValueError:
    print("Error: Invalid input!")
else:
    print(f"Division successful! Result: {result}")
finally:
    print("This block always executes, whether an exception occurs or not.")
```

### 06) Create a short program that prompts the user for a list of grades separated by commas. 
### Split the string into individual grades and use a list comprehension to convert each string to an integer. 
### You should use a try statement to inform the user when the values they entered cannot be converted.

```python
try:
    grades_input = input("Enter grades separated by commas: ")
    grades = [int(grade.strip()) for grade in grades_input.split(",")]
    print("Grades:", grades)
except ValueError:
    print("Error: Please enter valid integers separated by commas.")
```

### 07) WAP to create an udf divide(a,b) that handles ZeroDivisionError.

```python
def divide(a, b):
    try:
        result = a / b
        return f"Result: {result}"
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."

# Example usage
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print(divide(num1, num2))
```

### 08) WAP that gets an age of a person form the user and raises ValueError with error message: "Enter Valid Age" :
#### If the age is less than 18. 
#### otherwise print the age.

```python
class InvalidMarksError(Exception):
    pass

def calculate_grade(marks):
    if not (0 <= marks <= 100):
        raise InvalidMarksError("Marks should be between 0 and 100!")
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"

# Example usage
try:
    marks = int(input("Enter student marks (0-100): "))
    grade = calculate_grade(marks)
    print(f"Grade: {grade}")
except InvalidMarksError as e:
    print("Exception:", e)
except ValueError:
    print("Error: Please enter a valid number.")
```

### 09) WAP to raise your custom Exception named InvalidUsernameError with the error message : "Username must be between 5 and 15 characters long":
#### if the given name is having characters less than 5 or greater than 15.
#### otherwise print the given username.

```python
class InvalidUsernameError(Exception):
    """Custom exception for invalid usernames."""
    pass

def validate_username(username):
    try:
        if not (5 <= len(username) <= 15):
            raise InvalidUsernameError("Username must be between 5 and 15 characters long")
        print("Valid Username:", username)
    except InvalidUsernameError as e:
        print("Error:", e)

# Example usage
username = input("Enter a username: ")
validate_username(username)
```

### 01) WAP to print 1 to 10.

### 02) WAP to print 1 to n.

```python
n = int(input('Enter a last number of series:'));
for i in range(1,n+1):
    print(i);
```

### 03) WAP to print odd numbers between 1 to n.

```python
n = int(input('Enter a last number of odd series:'))
for i in range(1,n+1,2):
    print(i)
```

### 04) WAP to print numbers between two given numbers which is divisible by 2 but not divisible by 3.

```python
a = int(input('Enter first number:'))
b = int(input('Enter last number:'))
for i in range(a,b):
    if i%2==0 and i%3!=0:
        print(i)
```

### 05) WAP to print sum of 1 to n numbers.

### 06) WAP to print sum of series 1 + 4 + 9 + 16 + 25 + 36 + ...n.

```python
n = int(input('Enter last number of series:'))
sum = 0
for i in range(1,n+1):
    sum+=i**2
print(sum)
```

### 07) WAP to print sum of series 1 – 2 + 3 – 4 + 5 – 6 + 7 ... n.

```python
n = int(input('Enter last number of series:'))
sum = 0
sign = 1
for i in range(1,n+1):
    sum+=i*sign
    sign*=-1
print(sum)
```

### 11) WAP to find whether the given number is prime or not.

```python
n = int(input('enter a number:'))
count=0
if n>1:
    for i in range(2,n):
        if n%i==0:
            count+=1
    if count==0:
        print(n,'is prime number')
    else:
        print(n,'is not prime number')
else:
    print(n,' is not prime')
```

### 12) WAP to print sum of digits of given number.

```python
num = int(input('Eneter a number:'))
sum=0
while num!=0:
    sum+=num%10
    num//=10
print(sum)
```

### 13) WAP to check whether the given number is palindrome or not

```python
s = input('Eneter a number ')
if s.isnumeric():
    if s ==s[::-1]:
        print(s,'is palindrome')
    else:
        print(s,'is not palindrome')
else:
    print('enter a number')
```

### 14) WAP to print GCD of given two numbers.

```python
n1 = int(input('Enter a number:'))
n2 = int(input('Enter a number:'))

min = n1 if n1<n2 else n2
for i in range(2,min+1):
    if n1%i==0 and n2%i==0:
        gcd=i
print(f'gcd of {n1} and {n2} is {gcd}')
```

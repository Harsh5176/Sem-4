### 11) WAP to count number of vowels in given string using set.

```python
str1 = input('Enter the string:')
vowels = {'a','e','i','o','u'}
count=0

for i in str1:
    if vowels & set(i.lower()):
        count+=1
print('Number of vowels:',count)
```

### 12) WAP to check if a given string is binary string or not.

```python
st1 = input('Enter the string:')
binary={'0','1'}
s=set(st1)

if s <= binary:
    print('Binary String')
else:
    print('Not binary string')
```

### 13) WAP to sort dictionary by key or value.

```python
d = {21: 'watermelon', 43: 'banana', 3: 'apple'}
key1=list(d.keys())
value=list(d.values())
key1.sort()
value.sort()
d2={i:d[i] for i in key1}
print('sort by key :',d2)

temp = { d[i]:i for i in d }
keys = list(temp.keys())
keys.sort()
d2 = { temp[i] : i for i in keys }
print('Sort by value:',d2)
```

### 15) WAP to handle missing keys in dictionaries.
#### Example : Given, dict1 = {'a': 5, 'c': 8, 'e': 2} 
#### if you look for key = 'd', the message given should be 'Key Not Found', otherwise print the value of 'd' in dict1.

```python
dict1={'a':5,'c':8,'e':2}
key=input('Enter the key you want to print the value of:')
if key in list(dict1.keys()):
    print(dict1[key])
else:
    print('Key not found')
```
